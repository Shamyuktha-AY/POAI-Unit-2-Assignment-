```python
#3. Eight Puzzle — Greedy BFS

#For the 8-puzzle, the heuristic is usually the number of misplaced tiles.


import heapq

goal = "123456780"


def heuristic(state):
    count = 0

    for i in range(9):
        if state[i] != '0' and state[i] != goal[i]:
            count += 1

    return count


def greedy_bfs(start):
    pq = []
    heapq.heappush(pq, (heuristic(start), start))

    visited = set()

    while pq:
        h, state = heapq.heappop(pq)

        if state in visited:
            continue

        visited.add(state)

        print("State:", state, "Heuristic:", h)

        if state == goal:
            print("Goal reached")
            return

        zero = state.index('0')
        row = zero // 3
        col = zero % 3

        moves = []

        if row > 0:
            moves.append(zero - 3)

        if row < 2:
            moves.append(zero + 3)

        if col > 0:
            moves.append(zero - 1)

        if col < 2:
            moves.append(zero + 1)

        for new_pos in moves:
            board = list(state)

            board[zero], board[new_pos] = \
                board[new_pos], board[zero]

            new_state = ''.join(board)

            heapq.heappush(
                pq,
                (heuristic(new_state), new_state)
            )


start = "123405678"
greedy_bfs(start)
```

```python
                                                          #A* SEARCH


#1. A* Search — Tic-Tac-Toe

#For Tic-Tac-Toe, we can use A* to find the best next move. The heuristic estimates how good the board is.


import heapq

def heuristic(board):
    # Count possible winning lines for X
    lines = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]

    score = 0

    for a, b, c in lines:
        line = board[a] + board[b] + board[c]

        if "O" not in line:
            score += 1

    return -score


def astar(start):
    pq = []
    heapq.heappush(pq, (heuristic(start), 0, start))

    visited = set()

    while pq:
        f, g, board = heapq.heappop(pq)

        if board in visited:
            continue

        visited.add(board)

        print("Board:", board)
        print("g =", g, "h =", heuristic(board), "f =", f)

        if "_" not in board:
            print("Goal reached")
            return

        for i in range(9):
            if board[i] == "_":
                new_board = board[:i] + "X" + board[i+1:]

                new_g = g + 1
                new_h = heuristic(new_board)
                new_f = new_g + new_h

                heapq.heappush(
                    pq,
                    (new_f, new_g, new_board)
                )


start = "X_O__O___"
astar(start)
```

```python
#2. A* Search — Missionaries and Cannibals


import heapq

def heuristic(state):
    m, c, b = state

    # People remaining on left side
    return m + c


def valid(state):
    m, c, b = state

    mr = 3 - m
    cr = 3 - c

    if m < 0 or c < 0 or m > 3 or c > 3:
        return False

    if m > 0 and m < c:
        return False

    if mr > 0 and mr < cr:
        return False

    return True


def astar(start):
    pq = []

    # f, g, state, path
    heapq.heappush(pq, (heuristic(start), 0, start, []))

    visited = set()

    while pq:
        f, g, state, path = heapq.heappop(pq)

        if state in visited:
            continue

        visited.add(state)

        print("State:", state)
        print("g =", g, "h =", heuristic(state), "f =", f)

        if state == (0, 0, 1):
            print("Goal reached")
            print("Path:")

            for x in path + [state]:
                print(x)

            return

        m, c, b = state

        moves = [
            (1, 0),
            (2, 0),
            (0, 1),
            (0, 2),
            (1, 1)
        ]

        for dm, dc in moves:

            if b == 0:
                new_state = (m-dm, c-dc, 1)
            else:
                new_state = (m+dm, c+dc, 0)

            if valid(new_state):

                new_g = g + 1
                new_h = heuristic(new_state)
                new_f = new_g + new_h

                heapq.heappush(
                    pq,
                    (new_f, new_g, new_state,
                     path + [state])
                )


astar((3, 3, 0))
```

```python
                                                       # AO* SEARCH


#1. AO* — Tic-Tac-Toe


def heuristic(board):
    # X is better if it has more possible winning lines
    lines = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]

    score = 0

    for a, b, c in lines:
        line = [board[a], board[b], board[c]]

        if "O" not in line:
            score += 1

    return score


def ao_star(board):
    print("Board:", board)
    print("Heuristic:", heuristic(board))

    if "_" not in board:
        return board

    best_board = None
    best_h = -1

    for i in range(9):
        if board[i] == "_":

            new_board = board[:i] + "X" + board[i+1:]

            h = heuristic(new_board)

            if h > best_h:
                best_h = h
                best_board = new_board

    if best_board:
        return ao_star(best_board)


start = "X_O__O___"

result = ao_star(start)

print("Best path:")
print(result)
```

```python
#2. AO* — Missionaries and Cannibals


def heuristic(state):
    m, c, boat = state

    return m + c


def valid(state):
    m, c, boat = state

    mr = 3 - m
    cr = 3 - c

    if m < 0 or c < 0 or m > 3 or c > 3:
        return False

    if m > 0 and m < c:
        return False

    if mr > 0 and mr < cr:
        return False

    return True


def ao_star(state, path, visited):

    print("State:", state)
    print("Heuristic:", heuristic(state))

    if state == (0, 0, 1):
        print("Goal reached")
        return path + [state]

    if state in visited:
        return None

    visited.add(state)

    m, c, boat = state

    moves = [
        (1, 0),
        (2, 0),
        (0, 1),
        (0, 2),
        (1, 1)
    ]

    next_states = []

    for dm, dc in moves:

        if boat == 0:
            new_state = (m-dm, c-dc, 1)
        else:
            new_state = (m+dm, c+dc, 0)

        if valid(new_state):
            next_states.append(new_state)

    # Select state with lowest heuristic
    next_states.sort(key=heuristic)

    for new_state in next_states:

        result = ao_star(
            new_state,
            path + [state],
            visited
        )

        if result:
            return result

    return None


result = ao_star((3, 3, 0), [], set())

print("\nSolution:")
for state in result:
    print(state)
```

```python
                                             #MEMORY BOUND HEURISTIC SEARCH

#1. Memory-Bounded Search — Eight Puzzle


import heapq

goal = "123456780"
MAX_MEMORY = 10


def heuristic(state):
    count = 0

    for i in range(9):
        if state[i] != '0' and state[i] != goal[i]:
            count += 1

    return count


def get_moves(state):
    moves = []

    zero = state.index('0')
    row = zero // 3
    col = zero % 3

    if row > 0:
        moves.append(zero - 3)

    if row < 2:
        moves.append(zero + 3)

    if col > 0:
        moves.append(zero - 1)

    if col < 2:
        moves.append(zero + 1)

    return moves


def memory_bounded_search(start):

    pq = []
    heapq.heappush(pq, (heuristic(start), 0, start))

    visited = set()

    while pq:

        f, g, state = heapq.heappop(pq)

        if state in visited:
            continue

        visited.add(state)

        print("State:", state)
        print("g =", g, "h =", heuristic(state), "f =", f)
        print()

        if state == goal:
            print("Goal reached!")
            return

        for pos in get_moves(state):

            zero = state.index('0')

            board = list(state)
            board[zero], board[pos] = board[pos], board[zero]

            new_state = ''.join(board)

            if new_state not in visited:

                new_g = g + 1
                new_h = heuristic(new_state)
                new_f = new_g + new_h

                heapq.heappush(
                    pq,
                    (new_f, new_g, new_state)
                )

        # Limit memory
        if len(pq) > MAX_MEMORY:

            # Remove the worst node
            worst = max(pq)

            pq.remove(worst)
            heapq.heapify(pq)


start = "123405678"

memory_bounded_search(start)
```
